```python
def notificar(mensaje, titulo="ML Pipeline"):
    try:
        requests.post(
            "https://ntfy.sh/aaron-ml-experiments",
            data=mensaje.encode("utf-8"),
            headers={"Title": titulo, "Priority": "high"}
        )
    except Exception as e:
        print(f"Error al notificar: {e}")

acc = .68
f1 = .56
elapsed = 10010
# Ejemplo de uso después de entrenar
notificar(f"Modelo terminado — Accuracy: {acc:.4f}, F1: {f1:.4f}, Tiempo: {elapsed:.1f}s")
```

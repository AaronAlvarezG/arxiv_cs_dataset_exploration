<a id='error1'></a>
## 2. ❌ Error 1 — `AttributeError: module 'torch' has no attribute '_utils'`

### Síntoma

Al ejecutar:
```python
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
```

Se producía un traceback largo terminando en:
```
AttributeError: module 'torch' has no attribute '_utils'
```

El error ocurría **dentro de los módulos internos de PyTorch** (`torch._dynamo`, `torch._dynamo.variables.torch`), no en el código del usuario.

### Causa raíz

Instalación **corrupta o incompleta** de PyTorch. Esto ocurre cuando:
- La descarga se interrumpió parcialmente.
- Hay una **incompatibilidad de versiones** entre `torch`, `torchvision` y `torchaudio`.
- Se instalaron builds **nightly inestables** (ver Error 2).

El módulo interno `torch._utils` no estaba disponible porque la instalación era de una build de desarrollo rota.

### Solución

Desinstalar completamente y reinstalar desde cero con versiones pinned (ver sección 6).

---
<a id='error2'></a>
## 3. ❌ Error 2 — Versiones nightly inestables

### Síntoma

Al verificar el entorno se observaron versiones **inexistentes como releases estables**:

```
PyTorch     : 2.11.0+cu128   ← No existe como versión estable
CUDA        : 12.8
transformers: 5.5.0          ← No existe como versión estable
pandas      : 2.3.3          ← No existe como versión estable
```

### Causa raíz

El entorno fue construido con **builds nightly/dev** de PyTorch y Hugging Face.  
Los builds nightly usan números de versión adelantados a los releases y frecuentemente tienen **módulos internos rotos** entre builds consecutivas.

Adicionalmente, el índice nightly de PyTorch no garantiza estabilidad entre dependencias (`torch` + `torchvision` + `torchaudio` pueden ser de builds distintas).

### Solución

Reinstalar con versiones **pinned y estables**:  
- `transformers==4.47.1`
- `pandas==2.2.3`
- `gensim==4.3.3`
- `numpy==2.2.6`

> **Nota:** Para PyTorch, la versión estable `2.5.1` y `2.6.0` resultaron insuficientes por el problema de `sm_120` (ver Error 4). La solución definitiva requirió el nightly `cu128`, pero en este caso sí se usa el nightly de forma *controlada* con `--no-cache-dir`.

<a id='error3'></a>
## 4. ❌ Error 3 — Sintaxis de Windows CMD vs Linux

### Síntoma

Al intentar ejecutar el comando de instalación con continuación de línea (estilo Linux/bash):

```cmd
pip install torch==2.5.1 torchvision==0.20.1 torchaudio==2.5.1 \
    --index-url https://download.pytorch.org/whl/cu124
```

Se produjeron dos errores:
```
ERROR: Directory '\\' is not installable. Neither 'setup.py' nor 'pyproject.toml' found.
"--index-url" no se reconoce como un comando interno o externo.
```

### Causa raíz

En **Linux/macOS (bash/zsh)**, el carácter `\` al final de línea es una continuación de línea.  
En **Windows CMD**, `\` es el **separador de rutas de directorio**, no una continuación.  
CMD interpretó `\\` como una ruta de directorio y `--index-url` como un comando separado.

### Solución

En Windows CMD, escribir el comando **en una sola línea**:

```cmd
pip install torch==2.5.1 torchvision==0.20.1 torchaudio==2.5.1 --index-url https://download.pytorch.org/whl/cu124
```

> **Alternativa:** Si se usa **PowerShell**, la continuación de línea es el backtick `` ` ``:
> ```powershell
> pip install torch==2.5.1 torchvision==0.20.1 torchaudio==2.5.1 `
>     --index-url https://download.pytorch.org/whl/cu124
> ```

## 5. ❌ Error 4 — GPU no soportada: `sm_120` incompatible con PyTorch 2.5 y 2.6

### Síntoma

Incluso con PyTorch correctamente instalado (2.5.1 cu124 y 2.6.0 cu126), al importar torch aparecía:

```
UserWarning: NVIDIA GeForce RTX 5090 with CUDA capability sm_120 is not compatible
with the current PyTorch installation.
The current PyTorch install supports CUDA capabilities sm_50 sm_60 sm_61 sm_70
sm_75 sm_80 sm_86 sm_90.
```

### Causa raíz

La RTX 5090 usa la arquitectura **Blackwell** con compute capability `sm_120`.  
Las versiones estables de PyTorch hasta 2.6 solo incluyen kernels CUDA compilados hasta `sm_90` (arquitectura Hopper, e.g. A100/H100).

Cuando PyTorch encuentra una GPU con `sm_120`:
- Emite la advertencia mostrada arriba.
- **Hace fallback silencioso** — las operaciones parecen funcionar pero no usan la GPU realmente, o crashean con `CUDA error: no kernel image is available for execution on the device`.

```
Jerarquía de arquitecturas NVIDIA:
  sm_70  → Volta    (V100)         — 2017
  sm_75  → Turing   (RTX 2080)     — 2018
  sm_80  → Ampere   (A100, RTX 3090) — 2020
  sm_86  → Ampere   (RTX 3080)     — 2020
  sm_90  → Hopper   (H100)         — 2022
  sm_120 → Blackwell (RTX 5090)    — 2025  ← NO soportado en PyTorch ≤ 2.6 estable
```

### Por qué `cu124` y `cu126` no ayudan

El número `cu124`/`cu126` indica la **versión del toolkit CUDA** con que se compiló PyTorch, no las arquitecturas GPU soportadas.  
Lo determinante es la lista de `sm_XX` incluidos en la compilación.  
Ni `cu124` ni `cu126` incluyeron `sm_120` en sus builds estables de Windows.

### Solución

Instalar PyTorch **nightly con cu128**, que sí incluye kernels compilados para `sm_120` (ver sección 6).

<a id='solucion'></a>
## 6. ✅ Solución Definitiva — PyTorch Nightly cu128

### Explicación de cada parte del comando

A continuación se explica **cada flag** del comando de instalación:

```cmd
pip install --pre torch torchvision torchaudio --index-url https://download.pytorch.org/whl/nightly/cu128 --no-cache-dir
```

| Fragmento | Significado |
|---|---|
| `pip install` | Instala paquetes de Python |
| `--pre` | Incluye versiones pre-release y nightly (sin esto, pip solo busca versiones estables y no encontraría los builds cu128 con sm_120) |
| `torch torchvision torchaudio` | Los tres paquetes core de PyTorch (deben instalarse juntos para garantizar compatibilidad de versiones) |
| `--index-url https://download.pytorch.org/whl/nightly/cu128` | Usa el índice oficial de PyTorch para builds nightly con CUDA 12.8 (no están en PyPI estándar) |
| `--no-cache-dir` | Fuerza descarga fresca; evita que pip use en silencio una versión cacheada antigua incompatible |

### Paso 1 — Desinstalar la versión anterior

```cmd
pip uninstall torch torchvision torchaudio -y
```

- `-y` confirma automáticamente la desinstalación sin pedir confirmación interactiva.
- Es **crítico** desinstalar antes. Si se omite, pip puede reportar éxito sin realmente reemplazar los archivos.

### Paso 2 — Instalar PyTorch nightly cu128

```cmd
pip install --pre torch torchvision torchaudio --index-url https://download.pytorch.org/whl/nightly/cu128 --no-cache-dir
```

### Paso 3 — Instalar el resto del stack en versiones estables

```cmd
pip install transformers==4.47.1 pandas==2.2.3 gensim==4.3.3
```

- Fijar versiones explícitas (`==`) garantiza reproducibilidad del entorno.
- `numpy` se instala automáticamente como dependencia de torch/pandas.

### ¿Por qué nightly y no estable?

PyTorch 2.7 introdujo soporte para `sm_120` en los binarios, pero los wheels de Windows tuvieron retrasos en empaquetado y algunos kernels (como FlashAttention) estaban incompletos. Los builds nightly `cu128` son la opción más confiable en Windows con RTX 50-series hasta que salga un release estable completo (esperado en PyTorch 2.8+).

## 7. 🔍 Verificación Completa de la Instalación

Ejecuta las siguientes celdas en orden para verificar que todo funciona correctamente.

### 7.1 Versiones de paquetes

````python
import importlib

packages = {
    'torch': '2.7+ nightly cu128',
    'torchvision': '0.22+',
    'torchaudio': '2.7+',
    'transformers': '4.47.1',
    'pandas': '2.2.3',
    'numpy': '2.x',
    'gensim': '4.3.3',
}

print('=' * 55)
print(f'  {"Paquete":<20} {"Versión instalada":<20} {"Objetivo"}')
print('=' * 55)

all_ok = True
for pkg, target in packages.items():
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, '__version__', 'N/A')
        print(f'  {pkg:<20} {version:<20} {target}')
    except ImportError:
        print(f'  {pkg:<20} {"NO INSTALADO":<20} {target}')
        all_ok = False

print('=' * 55)
print('✅ Todos los paquetes importados correctamente.' if all_ok else '❌ Hay paquetes faltantes.')
````

### 7.2 Verificación de CUDA y GPU

````python
import torch

print('=' * 50)
print('  INFORMACIÓN DE PYTORCH Y CUDA')
print('=' * 50)

print(f'  PyTorch version : {torch.__version__}')
print(f'  CUDA version    : {torch.version.cuda}')

cuda_ok = torch.cuda.is_available()
print(f'  CUDA disponible : {"✅ Sí" if cuda_ok else "❌ No"}')

if cuda_ok:
    n_gpus = torch.cuda.device_count()
    print(f'  GPUs detectadas : {n_gpus}')
    for i in range(n_gpus):
        props = torch.cuda.get_device_properties(i)
        cap = torch.cuda.get_device_capability(i)
        vram_gb = props.total_memory / 1e9
        print(f'\n  GPU [{i}]: {props.name}')
        print(f'    VRAM              : {vram_gb:.1f} GB')
        print(f'    Compute Capability: sm_{cap[0]}{cap[1]}')
        
        # Verificar que es Blackwell (sm_120)
        if cap >= (12, 0):
            print(f'    Arquitectura      : ✅ Blackwell (sm_120) — soporte completo')
        elif cap >= (9, 0):
            print(f'    Arquitectura      : Hopper (sm_90)')
        elif cap >= (8, 0):
            print(f'    Arquitectura      : Ampere (sm_80/86)')
        else:
            print(f'    Arquitectura      : Generación anterior (sm_{cap[0]}{cap[1]})')
else:
    print('\n  ⚠️  CUDA no disponible. Posibles causas:')
    print('      - Driver NVIDIA no instalado')
    print('      - PyTorch instalado en versión CPU-only')
    print('      - GPU no detectada por el sistema')

print('=' * 50)
````

### 7.3 Test de cómputo real en GPU

> ⚠️ **Este es el test más importante.** `torch.cuda.is_available()` puede retornar `True` incluso cuando los kernels de `sm_120` no están presentes, porque solo verifica que el driver detecta una GPU CUDA. El verdadero test es ejecutar una operación en la GPU.

````python
import torch

print('TEST 1: Tensor simple en GPU')
print('-' * 40)
try:
    t = torch.tensor([2.0]).cuda()
    resultado = t * t
    assert resultado.item() == 4.0, 'Resultado incorrecto'
    print(f'  Entrada  : {t}')
    print(f'  t * t    : {resultado}')
    print(f'  Dispositivo: {resultado.device}')
    print('  ✅ PASSED — GPU ejecutando operaciones correctamente')
except Exception as e:
    print(f'  ❌ FAILED: {e}')
    print('  → Los kernels sm_120 no están presentes en esta build de PyTorch')
````

````python
import torch

print('TEST 2: Multiplicación de matrices (operación central en deep learning)')
print('-' * 60)
try:
    A = torch.randn(1024, 1024, device='cuda')
    B = torch.randn(1024, 1024, device='cuda')
    C = torch.matmul(A, B)
    torch.cuda.synchronize()  # Espera a que la GPU termine
    print(f'  Forma de A : {A.shape}')
    print(f'  Forma de B : {B.shape}')
    print(f'  Forma de C : {C.shape}')
    print(f'  Dispositivo: {C.device}')
    print(f'  Dtype      : {C.dtype}')
    print('  ✅ PASSED — MatMul 1024x1024 completado en GPU')
except Exception as e:
    print(f'  ❌ FAILED: {e}')
````

````python
import torch
import torch.nn as nn

print('TEST 3: Modelo simple + Optimizador (replica el error original del chat)')
print('-' * 60)
try:
    # Modelo mínimo
    model = nn.Linear(128, 10).cuda()
    
    # Esto era lo que causaba el AttributeError original
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    
    # Forward + backward pass
    x = torch.randn(32, 128, device='cuda')
    y = torch.randint(0, 10, (32,), device='cuda')
    
    criterion = nn.CrossEntropyLoss()
    output = model(x)
    loss = criterion(output, y)
    loss.backward()
    optimizer.step()
    
    print(f'  Modelo     : nn.Linear(128, 10) en cuda')
    print(f'  Optimizador: Adam (lr=1e-3)')
    print(f'  Loss       : {loss.item():.4f}')
    print('  ✅ PASSED — Forward + Backward + Optimizer step completados')
except Exception as e:
    print(f'  ❌ FAILED: {e}')
````

````python
import torch

print('TEST 4: Memoria GPU disponible')
print('-' * 40)
if torch.cuda.is_available():
    total = torch.cuda.get_device_properties(0).total_memory / 1e9
    reservada = torch.cuda.memory_reserved(0) / 1e9
    asignada = torch.cuda.memory_allocated(0) / 1e9
    libre = total - reservada
    
    print(f'  VRAM total    : {total:.2f} GB')
    print(f'  VRAM reservada: {reservada:.2f} GB')
    print(f'  VRAM asignada : {asignada:.2f} GB')
    print(f'  VRAM libre    : {libre:.2f} GB')
    print('  ✅ PASSED — Información de memoria accesible')
````

### 7.4 Diagnóstico final

````python
import torch
import importlib

print('=' * 55)
print('  DIAGNÓSTICO FINAL DEL ENTORNO')
print('=' * 55)

checks = []

# 1. PyTorch importable
try:
    import torch
    checks.append(('PyTorch importable', True, torch.__version__))
except Exception as e:
    checks.append(('PyTorch importable', False, str(e)))

# 2. CUDA disponible
cuda_avail = torch.cuda.is_available()
checks.append(('CUDA disponible', cuda_avail, torch.version.cuda))

# 3. sm_120 soportado
if cuda_avail:
    cap = torch.cuda.get_device_capability(0)
    sm120 = cap >= (12, 0)
    checks.append((f'Compute capability sm_120', sm120, f'sm_{cap[0]}{cap[1]} detectada'))

# 4. Operación GPU real
try:
    _ = torch.tensor([1.0]).cuda() * 2
    checks.append(('Cómputo real en GPU', True, 'OK'))
except Exception as e:
    checks.append(('Cómputo real en GPU', False, str(e)[:40]))

# 5. Adam optimizer
try:
    p = [torch.zeros(3, requires_grad=True)]
    _ = torch.optim.Adam(p, lr=1e-3)
    checks.append(('Adam optimizer', True, 'OK'))
except Exception as e:
    checks.append(('Adam optimizer', False, str(e)[:40]))

# 6. Paquetes clave
for pkg in ['transformers', 'pandas', 'gensim']:
    try:
        mod = importlib.import_module(pkg)
        checks.append((f'{pkg} importable', True, mod.__version__))
    except Exception as e:
        checks.append((f'{pkg} importable', False, str(e)[:40]))

# Mostrar resultados
all_passed = True
for name, ok, detail in checks:
    icon = '✅' if ok else '❌'
    print(f'  {icon} {name:<35} {detail}')
    if not ok:
        all_passed = False

print('=' * 55)
if all_passed:
    print('  🎉 ENTORNO COMPLETAMENTE FUNCIONAL')
    print('     La RTX 5090 está lista para entrenar modelos.')
else:
    print('  ⚠️  HAY PROBLEMAS — Revisar los items con ❌')
print('=' * 55)
````

## 8. 📝 Resumen de Comandos — Instalación desde cero
